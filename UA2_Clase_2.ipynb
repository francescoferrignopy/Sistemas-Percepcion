
# Visión por Computador — Teoría Completa (sin código)




## Visión por computador en pocas palabras

La **visión por computador** (*Computer Vision*, CV) es un campo de estudio que permite a un computador **analizar** y **entender** una imagen digital o un conjunto de imágenes, como por ejemplo un vídeo.

La idea principal de la CV —en conjunto con la robótica y otras disciplinas— es **mejorar** tareas que pueden ser **agotadoras o repetitivas** para los humanos. En los últimos años se han logrado muchos avances con la aparición de sistemas complejos de visión por computador y **aprendizaje profundo**, como las conocidas **redes neuronales convolucionales**. Estos avances han cambiado la forma de abordar numerosos problemas, como el **reconocimiento facial** o el análisis de **imágenes médicas**.




## Imágenes

Antes que nada, conviene entender **qué es** exactamente una imagen. De forma coloquial, podríamos describirla como una **representación visual** de algo que, por sí misma, incluye muchas características (color, formas, etc.). Para un computador, una imagen se describe mejor como una **matriz**, en la que cada valor es un **píxel**. Así, cuando hablamos de resolución **1080p**, nos referimos a una **matriz de 1080×1920 píxeles**.



## Color

<div style="text-align:center"><img src="cv_assets/Resources/Color_Channels.png" width="35%"/></div>

En una imagen en color hablamos de una **matriz tridimensional** donde cada dimensión corresponde a un **canal de color** (rojo, verde y azul). Las dimensiones de esta matriz variarán según el **espacio de color**, del que hablaremos más adelante.

Podemos describir una imagen de formas más complejas. Por ejemplo, la **formación del color** se debe principalmente a cómo la **luz** incide sobre la superficie de los objetos. Cuando vemos algo **negro**, en realidad percibimos **ausencia de luz**. La formación del color dependerá de la **longitud de onda** de los componentes principales de la luz blanca.

Si te gusta la física, te resultará interesante un fenómeno en el que puede apreciarse una **deformación del color**: las **estrellas**. En muchas fotografías del espacio, las formaciones rocosas lejanas se ven **rojizas**, mientras que las más cercanas tienden a **azuladas**. Este fenómeno fue descrito por el astrónomo estadounidense **Edwin Hubble** en 1929. Sabemos que el universo está en **expansión**; si el espacio se deforma, la **luz** que recibimos de esas estrellas también lo hace. Como consecuencia, su longitud de onda **aumenta** y el color percibido **se desplaza al rojo** en lugar de al azul.

<div style="text-align:center"><img src="cv_assets/Resources/nasa_Spiral.jpg" width="35%"/></div>

No entraremos mucho más en profundidad en teoría del color. La idea es comprender las **bases** de lo que trabajaremos en el resto del curso. Si te interesa, este tema es fascinante para investigarlo con mayor detalle.



## Idea general

Con la introducción a CV y un poco de contexto sobre formación de la imagen, pasamos a una de las tareas básicas: el **filtrado por color**. Filtrar color es **extraer** de una imagen la información correspondiente a un **color específico**. Antes de ello, veremos operaciones básicas de OpenCV (en el material original) para familiarizarnos con la librería y entender el código posterior.

Previo al filtrado, es esencial entender el concepto de **espacio de color**. Lo usaremos con frecuencia en el curso y te permitirá experimentar con espacios diferentes según la aplicación. Un **espacio de color** no es más que un **modelo tridimensional** que intenta describir la **percepción humana** del color; sus coordenadas definen un color concreto. Uno muy conocido es **RGB**, donde todos los colores se crean mezclando **rojo, verde y azul**. (En Python/OpenCV se usa a menudo **BGR**, que invierte el orden).

### Rango de tonos y ajuste de S/V

Para comprender mejor, considera este esquema aproximado de cómo se distribuyen los colores en el canal **Hue**:

<div style="text-align:center"><img src="cv_assets/Resources/hsv_colors.png" width="60%"/></div>

Por ejemplo, si buscamos **azul**, un rango de **Hue** podría ser **110–120**, o más amplio como **100–130** para abarcar variaciones. En ese caso los límites podrían escribirse (conceptualmente) como:

- Límite inferior: `H = 110`, `S = S_min`, `V = V_min`  
- Límite superior: `H = 120`, `S = S_max`, `V = V_max`

En cuanto a **S** y **V**:
- Cuanto **menor Saturación (S)** → más cercano a **blanco**.  
- Cuanto **menor Valor (V)** → más cercano a **negro**.

## Detección de bordes

La **detección de bordes** es fundamental en procesamiento de imágenes porque facilita el **reconocimiento de objetos**, la **segmentación** y muchas otras tareas. Los **bordes** son regiones donde existe un **cambio brusco** en los **niveles de gris**.

En esta parte se tratan dos enfoques clásicos: el **operador de Sobel** (basado en convoluciones) y el **detector de Canny** (que usa Sobel para estimar el gradiente).



### Operador de Sobel

El [**operador de Sobel**](https://www.researchgate.net/publication/239398674_An_Isotropic_3x3_Image_Gradient_Operator) se utiliza en procesamiento de imágenes, especialmente en algoritmos de **detección de bordes**. Calcula el **gradiente de intensidad** de la imagen en cada píxel mediante **convolución**. El resultado muestra los cambios de intensidad que pueden considerarse **bordes**. (La **convolución** es una operación matemática ampliamente usada en procesado de señales como **filtro**: combina dos funciones para obtener una tercera que representa cuánto modifica una a la otra).

### Detección de bordes de Canny

Para profundizar en Canny, consulta la [página de OpenCV](https://docs.opencv.org/trunk/da/d22/tutorial_py_canny.html).  
Es un **algoritmo multietapa** basado en Sobel. Sus fases principales son:

1. **Reducción de ruido** mediante un **filtro gaussiano**, definido por:

<div style="text-align:center"><img src="cv_assets/Resources/GaussFilter.png" width="50%"/></div>

donde cada valor del kernel viene dado por:

<div style="text-align:center"><img src="cv_assets/Resources/GaussDefinition.png" width="28%"/></div>

Este paso **suaviza** la imagen. Como Canny se basa en derivadas para estimar el gradiente (Sobel), es **sensible al ruido**; el suavizado reduce esa sensibilidad.

2. **Cálculo del gradiente**: se obtienen las derivadas **_Ix_** e **_Iy_** (mediante los Sobel horizontal y vertical). Con ellas se calcula la **magnitud** **_G_** y la **orientación** **_θ_** del borde:

<div style="text-align:center"><img src="cv_assets/Resources/magnitud_Orientation.png" width="28%"/></div>

3. **Supresión de no-máximos** (*Non-Maximum Suppression*): elimina píxeles que no sean máximos locales en la **dirección del gradiente**.

4. **Umbralización con histéresis** (*Hysteresis thresholding*): decide qué píxeles son realmente **bordes** y cuáles no, utilizando **dos umbrales** y la **conectividad**.  
Una explicación accesible de la histéresis puede encontrarse [aquí](https://towardsdatascience.com/canny-edge-detection-step-by-step-in-python-computer-vision-b49c3a2d8123).

## Transformaciones morfológicas

Las **transformaciones morfológicas** son, en mi opinión, de las operaciones más importantes en procesamiento de imágenes, ya que ayudan a **suprimir ruido** y realizar otras tareas. Son operaciones simples basadas en la **forma** de la imagen, y se aplican comúnmente sobre imágenes **binarias**. Funcionan con un **kernel** (elemento estructurante), por ejemplo una matriz de **5×5** de unos. Las cuatro transformaciones más comunes son:

- **Erosión**
- **Dilatación**
- **Apertura**
- **Cierre**

Para más información, consulta la [página de morfología de OpenCV](https://docs.opencv.org/trunk/d9/d61/tutorial_py_morphological_ops.html).
